```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler,MinMaxScaler,OneHotEncoder
#from sklearn.compose import ColumnTransformer
#from sklearn.pipeline import Pipeline
#from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest,f_regression,f_classif
#from sklearn.decomposition import PCA
```

```python
df=pd.read_csv("auto-mpg.csv")

display(df.head())
```

```python
print(df.dtypes)
```

```python
df.columns
```

```python
#To automatically find numeric columns
numeric_columns=df.select_dtypes(include=np.number).columns

print("Numeric Columns:")
print(numeric_columns)

#origin is often represented by numbers such as 1,2,3
#so technically pandas may identify it as numeric
#even though it represents categories.

```

```python
#To automatically find categorical columns:
categorical_columns=df.select_dtypes(include="object").columns

print("Categorical Columns:")
print(categorical_columns)
```

```python
df["origin"]=df["origin"].astype("object")
```

```python
#First Check missing values:
print(df.isnull().sum())

#Auto-MPG commonly represents missing horsepower values using?.
#So first convert ? into missing value:
```

```python
df["horsepower"]=pd.to_numeric(df["horsepower"].replace("?",np.nan),errors="coerce")

df["horsepower"]=df["horsepower"].replace(np.nan,df["horsepower"].mean())

print(df.isnull().sum())
```

```python
scaler=MinMaxScaler().set_output(transform="pandas")

scaled_data=scaler.fit_transform(df[numeric_columns])

scaled_data.head()
```

```python
#Suppose mpg is our target

X=df.drop("mpg",axis=1)
y=df["mpg"]
```

```python
#For a simple numerical example, we can select the best 5 features:

X_numeric=X.select_dtypes(include=np.number)

selector=SelectKBest(
    score_func=f_regression,
    k=5
).set_output(transform="pandas")

X_selected=selector.fit_transform(X_numeric,y)

print(X_selected.head())
```
